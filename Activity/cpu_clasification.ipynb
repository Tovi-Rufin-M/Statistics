# reading datafreme

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
df = pd.read_csv("../dataset/cpu_data.csv")
df
```

```python
# remove the data with 0 values
df = df[df["price"] != 0]
df
```

```python
# select only the necessary columns
df = df[["name", "price", "date"]]
```

```python
# showing the 4 degits only of the price
df_expensive = df[df["price"] > 1000]
df_expensive
```

```python
df_cheap = df[df["price"] < 50]
df_cheap
```

```python
df_cheap.describe()
```

# ploting the cheep and expencive cpu

```python
# showing the price over time for the cheap cpus
df_cheap["date"] = pd.to_datetime(df["date"])   # x-axis column
df_cheap["price"] = pd.to_numeric(df_cheap["price"], errors="coerce")  # y-axis column

df = df_cheap.dropna(subset=["date", "price"]).sort_values("date")

df_monthly = df.set_index("date").resample("M")["price"].mean().reset_index()

plt.figure(figsize=(10,5))
plt.bar(df["date"], df["price"], width=10)
plt.xlabel("Date")
plt.ylabel("Price")
plt.title("Price over Time")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df_expensive["price"].describe()
```

```python
# showing the price over time for the expensive cpus
df_expensive["date"] = pd.to_datetime(df_expensive["date"])   # x-axis column
df_expensive["price"] = pd.to_numeric(df_expensive["price"], errors="coerce")  # y-axis column

df = df_expensive.dropna(subset=["date", "price"]).sort_values("date")

plt.figure(figsize=(10,5))
plt.bar(df["date"], df["price"], width=10)
plt.xlabel("Date")
plt.ylabel("Price")
plt.title("Price over Time")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# showing the number of cpus developed each year
df["year"] = df["date"].dt.year
df_yearly = df.groupby("year").size().reset_index(name="count")
plt.figure(figsize=(10,5))
plt.bar(df_yearly["year"], df_yearly["count"], width=0.8)
plt.xlabel("Year")
plt.ylabel("Number of CPUs")
plt.title("Number of CPUs Developed Each Year")
```

```python
# number of cheep cpus developed each year
df_cheap["year"] = df_cheap["date"].dt.year
df_cheap_yearly = df_cheap.groupby("year").size().reset_index(name="count")
plt.figure(figsize=(10,5))
plt.bar(df_cheap_yearly["year"], df_cheap_yearly["count"], width=0.8)
plt.xlabel("Year")
plt.ylabel("Number of CPUs")
plt.title("Number of Cheap CPUs Developed Each Year")
```

```python
# number of expensive cpus developed each year
df_expensive["year"] = df_expensive["date"].dt.year
df_expensive_yearly = df_expensive.groupby("year").size().reset_index(name="count")
plt.figure(figsize=(10,5))
plt.bar(df_expensive_yearly["year"], df_expensive_yearly["count"], width=0.8)
plt.xlabel("Year")
plt.ylabel("Number of CPUs")
plt.title("Number of Expensive CPUs Developed Each Year")
```
